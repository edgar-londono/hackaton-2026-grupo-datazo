# Run the Datazo MVP

This notebook clones the public `main` branch, installs the small runtime dependency, and calls the same pipeline as the local command. It uses only the supplied challenge files.

```python
from pathlib import Path
import os
import subprocess

repository = Path('/content/hackaton-2026-grupo-datazo')
if not repository.exists():
    subprocess.run([
        'git', 'clone', '--depth', '1', '--single-branch', '--branch', 'main',
        'https://github.com/edgar-londono/hackaton-2026-grupo-datazo.git',
        str(repository),
    ], check=True)

project = repository / 'grupo-datazo'
os.chdir(project)
print(f'Project folder: {project}')
```
